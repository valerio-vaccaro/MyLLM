## 1. The whole course in one diagram

```
 text files ─clean─► corpus ─tokenize─► ids ─split─► train / validation
                                              │
                                    random windows (x, y = x shifted)
                                              ▼
              ┌──────────── GPT: embeddings → N × (attention + MLP) → logits ───────────┐
              │ loss = −log p(true next token) ; AdamW ; warm-up + cosine LR ; clip      │  training
              └──────────────────────────────────────────────────────────────────────────┘
                                              │ best checkpoint (weights + config + vocab)
                          ┌───────────────────┴───────────────────┐
                    evaluate: loss, perplexity, accuracy     generate: temperature, top-k, top-p
```

## 2. Glossary

| term | meaning |
|------|---------|
| **token** | the unit the model reads: here a lower-case word or punctuation mark |
| **vocabulary** | the list of known tokens; id = position in the list |
| **embedding** | a learned vector representing a token (or a position) |
| **logits** | raw scores for every vocabulary word, before softmax |
| **softmax** | turns scores into probabilities that sum to 1 |
| **attention** | each token takes a weighted average of information from other tokens |
| **causal mask** | forbids attending to future tokens |
| **block size / context** | how many tokens the model sees at once |
| **parameter** | a learned number (weight) |
| **loss (cross-entropy)** | `−log` probability given to the true next token |
| **perplexity** | `exp(loss)`; effective number of equally likely choices |
| **epoch** | one pass over the training data |
| **overfitting** | train loss falls but validation loss rises: memorising instead of learning |
| **temperature** | divides logits before softmax: low = safe, high = random |
| **top-k / top-p** | restrict sampling to the likeliest tokens |
| **checkpoint** | saved model file |

## 3. Quiz

Answer first, then open the solution.

**Q1.** An untrained model with a 4,000-word vocabulary has a loss of about…?
<details><summary>Answer</summary>log(4000) ≈ 8.3 (perplexity 4000): it is guessing uniformly.</details>

**Q2.** Why is `y` the input shifted by one position?
<details><summary>Answer</summary>At every position the training target is the <em>next</em> token, so one window gives block_size training examples.</details>

**Q3.** What would happen without the causal mask?
<details><summary>Answer</summary>The model could read the answer (the next token) from the input, reach near-zero loss in training and be useless for generation.</details>

**Q4.** Training loss keeps falling but validation loss rises. What is happening and what can you do?
<details><summary>Answer</summary>Overfitting. More data, dropout/weight decay, a smaller model or earlier stopping (we save the <em>best</em> validation checkpoint).</details>

**Q5.** Why does the vocabulary dominate the parameters of our model?
<details><summary>Answer</summary>The embedding table is vocab × n_embd (4000 × 128 = 512k of ~1.1M); transformer layers are small at width 128.</details>

**Q6.** You want more creative text. Which knob and in which direction?
<details><summary>Answer</summary>Raise the temperature (and/or top-p); the cost is more mistakes.</details>

**Q7.** Why do we compare perplexity with a unigram baseline?
<details><summary>Answer</summary>To show that the model uses <em>context</em>, not only word frequencies.</details>

## 4. Project ideas

| level | project |
|-------|---------|
| ★ | Train on your own texts (emails, a novel, source code) and compare the style of the samples |
| ★ | Add a `--min-p` or repetition penalty to `generate.py` and show the effect on loops |
| ★★ | Replace the word tokenizer with a **character-level** one; compare loss per character and sample quality |
| ★★ | Add an end-of-text token to training data so the model learns to stop |
| ★★★ | Implement **byte-pair encoding** (subword) and compare with word-level at equal parameter count |
| ★★★ | Add a learned-position → **rotary** position encoding and measure the change |

## 5. Where limits come from (and how real LLMs go beyond)

| this course | production LLMs |
|-------------|-----------------|
| ~1M parameters | 10⁹–10¹² parameters |
| ~1M tokens of text | 10¹²+ tokens |
| word-level vocabulary 4k | subword vocabulary 30k–200k |
| context 64 | context 8k–1M |
| continues text only | + instruction tuning and human-feedback training to *answer* |
| 10 minutes on a CPU | weeks on thousands of GPUs |

The **algorithm is the same**; the scale is what changes.
